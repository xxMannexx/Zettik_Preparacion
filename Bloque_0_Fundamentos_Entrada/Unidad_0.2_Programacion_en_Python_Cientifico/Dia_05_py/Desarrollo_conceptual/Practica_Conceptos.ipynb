6.1 Definir una clase con atributos de instancia

```python
class Vector:                  ## Creamos la clase con atributos de instancia
    def __init__(self, x, y):
        self.x = x
        self.y = y

v = Vector(3, 4)
print(v.x)
print(v.y)
print(v.__dict__)  ## Muestra el diccionario de nombres con sus valores
```

6.2 Distinguir los tipos de metodos


```python
class Util:
    base = 10

    def metodo(self): return "Instancia"

    @classmethod
    def de_clase(cls):
        return f"clase, base={cls.base}"

    @staticmethod
    def estatico(): return "estatico"

u = Util()
print(u.metodo(), u.de_clase(), u.estatico())
```

6.3 Heredar y sobrescribir con super()

```python
class Base:
    def __init__(self,n):
        self.n = n

    def info(self):
        return f"base: {self.n}"

class Sub(Base):
    def __init__(self,n,extra):
        super().__init__(n) ## # extiende el __init__ de Base
        self.extra = extra

    def info(self): return f"sub {self.n}/{self.extra}"

s = Sub(3,4)
print(s.info(),[c.__name__ for c in Sub.__mro__])
```

6.4 Escribir una funcion polimorfica (duck typing)

```python
class Circulo:
    def area(self):
        return 3.14159 * 4
class Cuadrado:
    def area(self):
        return 16

def area_total(*figuras):
    ## Polimorfica:no conoce los tipos
    return sum(f.area() for f in figuras)

print(area_total(Circulo(),Cuadrado()))
```

6.5 Integrar mi propio modelo de datos
## Hacer el objeto idiomatico

```python
class Coleccion:
    def __init__(self,*items):
        self.items = list(items)

    def __len__(self):
        return len(self.items)

    def __getitem__(self,i):
        return self.items[i]

    def __repr__(self):
        return f"Coleccion({self.items})"

c = Coleccion(10,20,30)
print(len(c))
print(c[1])
print(c)

for x in c:
    print(x, end= " ")
```
